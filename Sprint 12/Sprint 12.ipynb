### 1. 
Consulta la API pública JSONPlaceholder utilizando el método GET para obtener:

* Lista de publicaciones
* Lista de usuarios (/users)
* Lista de tareas (/todos)

```python
BASE= "https://jsonplaceholder.typicode.com" # URL base


endpoints = {"Posts": f"{BASE}/posts", # diccionario que contiene los endpoints a consultar
            "Users": f"{BASE}/users",
            "ToDos": f"{BASE}/todos"}

data = {} # diccionario para almacenar las respuestas JSON

for nombre, enlace in endpoints.items(): # recorrer cada endpoint
    GET = requests.get(enlace) # hacer una llamada GET 
    
    json_data = GET.json() # convertir a JSON la información recibida
    
    data[nombre] = json_data # guardar el resultado en el diccionario
    
    print(f"{nombre}") # nombre del endpoint
    print(f"URL: {enlace}") # enlace para acceder al endpoint
    print(f"Primer elemento de {nombre}:")
    print(json_data[0])
    print("_" * 200)
```

### 2. 
Muestra por pantalla:

* La cantidad total de cada recurso.
* El código de estado de cada petición.

```python
for nombre, enlace in endpoints.items(): # recorrer cada endpoint
    GET = requests.get(enlace)
    
    estado = GET.status_code # recibir el código de estado
    json_data = GET.json()
    cantidad_items = len(json_data) # cantidad total de elementos de endpoint
    
    print(f"Recurso: {nombre}")
    print(f"URL consultada: {enlace}")
    print(f"Código de estado HTTP: {estado}")
    print(f"Cantidad total de elementos: {cantidad_items}")
    print("_" * 200)
```

### 3. 
Realiza una petición a una publicación inexistente para obtener un error 404 y muestra el código de estado recibido.

```python
inexistente=f"{BASE}/posts/inexistente"

GET = requests.get(inexistente) # petición GET a un recurso inexistente 

estado = GET.status_code # solicitar el código de estado

print(f"URL: {inexistente}")
print(f"Código de estado HTTP: {estado}")
```

### 4. 
Realiza una petición POST para crear una nueva publicación ficticia. Incluye el título, el cuerpo del mensaje y un userId.

* Muestra la respuesta JSON.
* Muestra el código de estado.

```python
posts = f"{BASE}/posts"

payload = {"title": "Post 1",
            "body": "Blah blah blah.. ",
            "userId": 1}

POST = requests.post(posts, json=payload) # petición POST (URL + diccionario)

status_code = POST.status_code # el código de estado

response_json = POST.json() # convertir la respuesta a JSON

print("Petición POST para crear una nueva publicación ficticia")
print("_" * 200)
print(f"URL: {posts}")
print("Datos enviados:")
print(payload)
print("_" * 200)
print("Respuesta JSON recibida:")
print(response_json)
print("_" * 200)
print(f"Código de estado HTTP: {status_code}")
```

### 5. 
Realiza una petición PATCH para modificar parcialmente una publicación existente.

* Muestra la respuesta JSON.
* Muestra el código de estado.

```python
post_id = 1
enlace_post = f"{BASE}/posts/{post_id}"

payload = {"title": "Título modificado con PATCH"} # el nuevo título

PATCH = requests.patch(enlace_post, json=payload) # realizar la petición PATCH (enlace + cambio)

estado = PATCH.status_code # obtener el código de estado

PATCH_json = PATCH.json() # convertir la respuesta a JSON

print("Petición PATCH para modificar parcialmente una publicación")
print(f"URL: {enlace_post}")
print("Datos enviados:")
print(payload)
print("_" * 200)
print("Respuesta JSON recibida:")
print(PATCH_json)
print("_" * 200)
print(f"Código de estado HTTP: {estado}")
```

### 6. Haz una petición DELETE sobre una publicación.

* Muestra la respuesta JSON.
* Muestra el código de estado.

```python
post_id = 1
url_post = f"{BASE}/posts/{post_id}"

DELETE = requests.delete(url_post)# realizar la petición DELETE

estado = DELETE.status_code # obtener el código de estado

DELETE_json = DELETE.json() # convertir a JSON

print("Petición DELETE para eliminar una publicación")
print(f"URL: {url_post}")
print("_" * 200)
print("JSON recibido:")
print(DELETE_json)
print("_" * 200)
print(f"Código de estado HTTP recibido: {estado}")
```

### 1. 
Utiliza la API de Open Data BCN para:

* Buscar un dataset de tu interés mediante package_search y package_show.

> *Open Data BCN* emplea el software CKAN permite buscar y obtener detalles de conjuntos de datos mediante los endpoints package_search y package_show.

> He realizado una búsqueda usando la palabra clave *"aeroport"* para encontrar datasets relacionados con el aeropuerto Josep Tarradellas Barcelona‑El Prat.


> Los endpoints usados son:
> - package_search: para buscar datasets por texto
> - package_show: para obtener información completa de un dataset a partir de su ID

```python
BASE = "https://opendata-ajuntament.barcelona.cat/data/api/3/action" # enlace de Open Data BCN

# 1) Búsqueda de datasets que contengan "aeroport"
palabra = "aeroport"
enlace = f"{BASE}/package_search?q={palabra}"

busqueda = requests.get(enlace) # petición GET

print("Código de estado de package_search:", busqueda.status_code) # código de estado
busqueda_json = busqueda.json() # cambio a JSON

total_resultados = busqueda_json["result"]["count"] # número total de resultados
print("Total de resultados encontrados:", total_resultados)

primer_dataset = busqueda_json["result"]["results"][0] # el primer dataset de resultados
dataset_id = primer_dataset["id"]
dataset_nombre = primer_dataset.get("name", "")

print("ID del primer dataset encontrado:", dataset_id)
print("Nombre del primer dataset:", dataset_nombre)

print("\nPrimer dataset (resumen):")
print(primer_dataset) # ejemplo del dataset

# 2) Obtener los detalles completos con package_show
show_url = f"{BASE}/package_show?id={dataset_id}"
respuesta_show = requests.get(show_url)

print("\nStatus code de package_show:", respuesta_show.status_code)

show_json = respuesta_show.json()

print("\nDetalles completos del datase:")
print(show_json)
```

### 3. 
Con el resource_id del recurso seleccionado, realiza una consulta mediante datastore_search para:

* Recuperar al menos 100 registros.

```python
BASE = "https://opendata-ajuntament.barcelona.cat/data/api/3/action/datastore_search" # endpoint base de Open Data BCN

RESOURCE_ID = "daaf92f4-8879-40f0-bc35-d550f1f9582a" # CSV del aeropuerto 2025

params = {"resource_id": RESOURCE_ID, # parámetros de la consulta: id
          "limit": 100}  # 100 registros

respuesta = requests.get(BASE, params=params) # petición GET

print("Status code:", respuesta.status_code) # el código de estado

data = respuesta.json() # convertir a JSON

print("Success:", data.get("success")) # comprobar si la API ha respondido correctamente a nivel lógico

registros = data["result"]["records"] # extraer los registros

print("Número de registros recuperados:", len(registros)) # cantidad de registros recuperados

print("\nPrimer registro:")
print(registros[0])
```

### 4. 
Convierte los resultados a un DataFrame.

```python
df = pd.DataFrame(registros) # convertir la lista en un DataFrame

print("Primeras filas del DataFrame:")
display(df.head())

print("\nDimensiones del DataFrame (filas, columnas):", df.shape)
print("\nNombres de columnas:")
print(df.columns)
```

### 5. 
Guarda DataFrame en un archivo .csv.

```python
archivo = "aeropuerto_bcn_dataset.csv" # nombre

df.to_csv(archivo, index=False, encoding="utf-8") # guardar el Df en un archivo CSV sin índice y con la codificación que asegura la correcta representación de caracteres especiales

print(f"Archivo CSV guardado correctamente con el nombre: {archivo}")
```
